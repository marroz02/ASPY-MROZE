```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Definiujemy 3 wierzchołki głównego trójkąta
wierzcholki = np.array([[0, 0], [1, 0], [0.5, np.sqrt(3) / 2]])

# 2. Punkt startowy i liczba kroków
punkt = np.array([0.5, 0.5])
kroki = 15000
punkty = np.zeros((kroki, 2))

# 3. Gra w chaos: w każdym kroku losujemy wierzchołek i przesuwamy się w połowie drogi do niego
rng = np.random.default_rng()
losowe_wierzcholki = wierzcholki[rng.integers(0, 3, size=kroki)]

for i in range(kroki):
  punkt = (punkt + losowe_wierzcholki[i]) / 2
  punkty[i] = punkt

# 4. Rysujemy wynik
plt.figure(figsize=(7, 6))
plt.scatter(punkty[:, 0], punkty[:, 1], s=0.2, color='royalblue')
plt.title('Trójkąt Sierpińskiego (Gra w Chaos)')
plt.axis('off')  # Ukrywamy osie dla lepszego efektu
plt.show()
```
