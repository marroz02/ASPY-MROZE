```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Generujemy kąt t od 0 do 12*pi (6 pełnych obrotów)
t = np.linspace(0, 12 * np.pi, 10000)

# 2. Równanie parametryczne Krzywej Motyla
r = (
    np.exp(np.sin(t))
    - 2 * np.cos(4 * t)
    + np.sin((2 * t - np.pi) / 24) ** 5
)

x = r * np.cos(t)
y = r * np.sin(t)

# 3. Rysowanie z gradientem kolorów
plt.figure(figsize=(8, 8))
plt.scatter(x, y, c=t, cmap='plasma', s=1)  # Kolor zmienia się wraz z kątem t
plt.title('Krzywa Motyla (Temple H. Fay)')
plt.axis('equal')  # Zachowanie proporcji osi
plt.axis('off')  # Ukrycie ramek dla lepszego efektu wizualnego
plt.show()
```
