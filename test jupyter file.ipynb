```python
import numpy as np
import matplotlib.pyplot as plt
# Losowanie 1000 wartości z rozkładu normalnego (średnia 0, odchylenie standardowe 1)
dane = np.random.randn(1000)
# Tworzenie histogramu
plt.hist(dane, bins=30, color='blue', edgecolor='black')
# Dodanie tytułu i etykiet osi
plt.title('Histogram losowo wygenerowanych danych')
plt.xlabel('Wartość')
plt.ylabel('Częstość')
# Pokazanie wykresu
plt.show()
```
